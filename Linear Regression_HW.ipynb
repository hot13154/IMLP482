## 用訓練資料來 fit 函數
1. 只用訓練資料集的資料進行linear regression演算法<br>
2. 並計算出訓練階段的MSE
3. 畫出目標點(藍色)與預測點(紅色)的對應圖
![](images/LR_train.PNG)

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

X_train = X_train.reshape(-1, 1)
#只用訓練資料集的資料進行linear regression演算法
regr_1 = LinearRegression()

regr_1.fit(X_train,Y_train)

Y_pred = regr_1.predict(X_train)
#並計算出訓練階段的MSE
mse = mean_squared_error(Y_train, Y_pred)
print("Training MSE =", mse)
#畫出目標點(藍色)與預測點(紅色)的對應圖 
W = regr_1.coef_
b = regr_1.intercept_

plt.scatter(X_train, Y_train, color='blue', label='Target Points')
plt.plot(X_train, Y_pred, color='red', label='Predicted Line')

plt.xlabel('X_train')
plt.ylabel('Y_train')
plt.title('Linear Regression - Training Data')
plt.legend()
plt.show()
```
